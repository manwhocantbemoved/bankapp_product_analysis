```python
categories = {
    "Login/Authentication": [
        "login", "log in", "password", "otp", "verification", "locked", "biometrics",
        "sign in", "register",
        "hindi makapasok", "ayaw mag-open", "hindi ma-access", "maka open", "di ma access", "ma-access"
    ],
    "App Stability": [
        "crash", "closes", "freeze", "white screen", "blank screen", "not responding",
        "not working", "maintenance", "bug", "glitch", "broken", "error", "downtime",
        "hindi gumagana", "di gumagana", "sira", "bugok", "puro error"
    ],
    "Performance": [
        "slow", "lag", "loading", "takes forever", "takes a long time",
        "mabagal", "matagal", "pahirap"
    ],
    "Transaction Failures": [
        "something went wrong", "transaction failed", "unable to transfer", "transfer failed",
        "unable to send", "cannot process", "unsuccessful", "cant pay", "instapay",
        "hindi maka send", "hindi ma-transfer", "di ma transfer"
    ],
    "Customer Service": [
        "customer service", "no response", "no reply", "support ticket", "complaint",
        "not assisting", "poor response", "help center",
        "walang sagot", "walang response", "sumagot"
    ],
}
```

```python
def tag_complaint(text):
    text = text.lower()
    tags = [cat for cat, kws in categories.items() if any(kw in text for kw in kws)]
    return tags if tags else ["Uncategorized"]
```

```python
test_cases = [
    ("The app keeps crashing every time I open it", ["App Stability"]),
    ("hindi ko ma-access yung account ko", ["Login/Authentication"]),
    ("so slow, matagal mag-load", ["Performance"]),
    ("something went wrong, cannot transfer money", ["Transaction Failures"]),
    ("no response from customer service", ["Customer Service"]),
    ("great app, works perfectly", ["Uncategorized"]), 
]

for text, expected in test_cases:
    result = tag_complaint(text)
    status = "PASS" if result == expected else "FAIL"
    print(f"{status} | text: '{text}' | expected: {expected} | got: {result}")
```

```python
df_clean['complaint_tags'] = df_clean['content'].apply(tag_complaint)
```

```python
tag_counts_by_bank = (
    df_clean.explode('complaint_tags')
    .groupby(['bank', 'complaint_tags'])
    .size()
    .unstack(fill_value=0)
)
print(tag_counts_by_bank)
```
